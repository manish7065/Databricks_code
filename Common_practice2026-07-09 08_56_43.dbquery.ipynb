```sql


-- revenue and profit by region and item type

select * from workspace.my_db.delta_sales;

select
sum(`Total Revenue`) as revenue,
sum(`Total Profit`) as profit,
Region,
`Item Type`
from
workspace.my_db.delta_sales
group by Region, `Item Type`
ORDER BY revenue DESC;
--Table name: workspace.)


--- monthly trend
SELECT date_trunc('month', to_date(`Order Date`, 'M/d/yyyy')) AS  month, sum(`Total Revenue`) AS revenue FROM workspace.my_db.delta_sales GROUP BY 1 ORDER BY 1;


select `Item Type`,
SUM(`Total Profit`)/ SUM(`Total Revenue`) * 100 as profit_margin
from workspace.my_db.delta_sales
group by `Item Type`
order by profit_margin desc;

SELECT Country,
`Order Date` ,
SUM(`Total Revenue`) over(PARTITION BY COUNTRY ORDER BY `Order Date`) AS TOTAL_REVENUE
FROM workspace.my_db.delta_sales;




```
